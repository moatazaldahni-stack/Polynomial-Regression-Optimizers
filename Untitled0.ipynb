```python
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
m = 500
X = np.random.rand(m, 1) * 4 - 2
y = 2 * (X**2) + 3 * X + 1 + np.random.randn(m, 1) * 0.5


X_poly = np.c_[np.ones((m, 1)), X, X**2]

epochs = 50
lr = 0.05
momentum = 0.9

w = np.random.randn(3, 1) * 0.01
v = np.zeros_like(w)
losses = []

for epoch in range(epochs):
    predictions = X_poly.dot(w)
    error = predictions - y

    grad = (2/m) * X_poly.T.dot(error)

    v = momentum * v + lr * grad
    w = w - v

    losses.append(np.mean(error**2))

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(losses, color='blue', linewidth=2)
plt.title('Cost Curve - GD with Momentum')
plt.xlabel('Epochs')
plt.ylabel('MSE')

plt.subplot(1, 2, 2)
sorted_idx = np.argsort(X.flatten())
plt.scatter(X, y, color='gray', alpha=0.5, label='Data')
plt.plot(X[sorted_idx], X_poly[sorted_idx].dot(w), color='red', linewidth=3, label='Predicted Curve')
plt.title('Polynomial Regression Prediction')
plt.legend()

plt.show()
```
